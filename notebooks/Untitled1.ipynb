```python
import numpy as np
import matplotlib.pyplot as plt

# generate skewed data
np.random.seed(0)
data = np.random.lognormal(mean=1, sigma=0.5, size=1000)

# Count histogram
plt.figure()
plt.hist(data, bins=20)
plt.title("Histogram (Count / Frequency)")
plt.xlabel("Value")
plt.ylabel("Count")

plt.show()

# Density histogram
plt.figure()
plt.hist(data, bins=20, density=True)
plt.title("Histogram (Density)")
plt.xlabel("Value")
plt.ylabel("Density")

plt.show()

```
